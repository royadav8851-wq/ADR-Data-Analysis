```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    "Report_ID": [
        "ADR001", "ADR002", "ADR003", "ADR004", "ADR005",
        "ADR006", "ADR007", "ADR008", "ADR009", "ADR010",
        "ADR011", "ADR012"
    ],
    "Age": [45, 62, 28, 70, 34, 50, 41, 23, 66, 39, 55, 31],
    "Gender": [
        "Female", "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female", "Male"
    ],
    "Suspected_Drug": [
        "Amoxicillin", "Diclofenac", "Carbamazepine", "Metformin",
        "Allopurinol", "Amoxicillin", "Diclofenac", "Metformin",
        "Allopurinol", "Carbamazepine", "Amoxicillin", "Diclofenac"
    ],
    "ADR": [
        "Rash", "Gastric Ulcer", "Dizziness", "Lactic Acidosis",
        "SJS_TEN", "Diarrhea", "Stomach Pain", "Nausea",
        "Rash", "Dizziness", "Diarrhea", "Gastric Ulcer"
    ],
    "Severity": [
        "Moderate", "Severe", "Mild", "Severe", "Severe", "Mild",
        "Moderate", "Mild", "Moderate", "Mild", "Mild", "Severe"
    ],
    "Onset_Days": [2, 5, 3, 7, 10, 1, 4, 2, 6, 3, 2, 5],
    "Outcome": [
        "Recovered", "Recovered", "Recovering", "Recovered",
        "Recovering", "Recovered", "Recovered", "Recovering",
        "Recovered", "Recovered", "Recovering", "Recovered"
    ],
    "System_Organ_Class": [
        "Skin Disorders", "Gastrointestinal Disorders",
        "Nervous System Disorders", "Metabolic Disorders",
        "Skin Disorders", "Gastrointestinal Disorders",
        "Gastrointestinal Disorders", "Gastrointestinal Disorders",
        "Skin Disorders", "Nervous System Disorders",
        "Gastrointestinal Disorders", "Gastrointestinal Disorders"
    ]
}

df = pd.DataFrame(data)

display(df)
print("ADR dataset created successfully!")
```

```python
print("Total ADR Reports:", len(df))
print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
df.info()
```

```python
soc_counts = df["System_Organ_Class"].value_counts()

plt.figure(figsize=(10, 6))
sns.barplot(x=soc_counts.values, y=soc_counts.index)

plt.title("ADR by System Organ Class")
plt.xlabel("Number of Reports")
plt.ylabel("System Organ Class")
plt.tight_layout()
plt.show()
```

```python
severity_counts = df["Severity"].value_counts()

plt.figure(figsize=(7, 5))
sns.barplot(x=severity_counts.index, y=severity_counts.values)

plt.title("ADR Severity Analysis")
plt.xlabel("Severity")
plt.ylabel("Number of Reports")
plt.tight_layout()
plt.show()

display(severity_counts)
```

```python
drug_counts = df["Suspected_Drug"].value_counts()

plt.figure(figsize=(9, 5))
sns.barplot(x=drug_counts.values, y=drug_counts.index)

plt.title("ADR Reports by Suspected Drug")
plt.xlabel("Number of Reports")
plt.ylabel("Suspected Drug")
plt.tight_layout()
plt.show()
```

```python
symptom_counts = df["ADR"].value_counts()

plt.figure(figsize=(9, 6))
sns.barplot(x=symptom_counts.values, y=symptom_counts.index)

plt.title("Commonly Reported Adverse Drug Reactions")
plt.xlabel("Number of Reports")
plt.ylabel("Adverse Drug Reaction")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Onset_Days",
    bins=7,
    kde=False
)

plt.title("ADR Onset Timeline")
plt.xlabel("Days Until Reaction Onset")
plt.ylabel("Number of Reports")
plt.tight_layout()
plt.show()

print("Average onset:", round(df["Onset_Days"].mean(), 2), "days")
```

```python
outcome_counts = df["Outcome"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    outcome_counts.values,
    labels=outcome_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("ADR Patient Outcomes")
plt.show()

display(outcome_counts)
```
